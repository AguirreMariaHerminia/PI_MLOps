Concateno de los df

```python
plataformas = (pd.concat([dfamz,dfdis,dfhul,dfnet], axis=0))
plataformas.shape
```

Reemplazo en la columna 'Rating' los valores nulos por 'G'

```python
plataformas['rating'] = plataformas['rating'].fillna('G')
plataformas
```

Cambio el formato del campo 'data_added'

```python
plataformas['date_added'] = pd.to_datetime(plataformas['date_added'])
plataformas
```

Reemplazo las letras mayusculas por minusculas

```python
plataformas['title']=plataformas['title'].str.lower()
plataformas['type']=plataformas['type'].str.lower()
plataformas['director']=plataformas['director'].str.lower()
plataformas['cast']=plataformas['cast'].str.lower()
plataformas['country']=plataformas['country'].str.lower()
plataformas['listed_in']=plataformas['listed_in'].str.lower()
plataformas['description']=plataformas['description'].str.lower()
plataformas.tail()
```

Divido el campo duracion para poder identificar las peliculas de las series

```python
div = plataformas["duration"].str.split(' ', n=1, expand=True)
plataformas["duration_int"] = div[0]
plataformas["duration_type"] = div[1]
plataformas.head(1)
```

```python
plataformas["duration_int"] = plataformas["duration_int"].fillna(0)
plataformas['duration_int'] = plataformas['duration_int'].astype('int64')
```

```python
plataformas = plataformas.fillna('Sin dato')
plataformas.isna().sum()
```

Calculo la media de las valoraciones, agrupando por "movieId"

```python
mean_score= ratings.groupby('movieId')['rating'].mean()
```

Elimino los duplicados

```python
ratings =ratings.drop_duplicates(subset='movieId')
```

Agrego la columna 'score' con la valoracion de los usuarios de cada pelicula

```python
plataformas['score'] = ratings.groupby('movieId')['rating'].transform(lambda x: mean_score.loc[x.name])
```
