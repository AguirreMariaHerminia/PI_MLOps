Funcion 1: devulve la pelicula con mayor duracion del año indicado segun cada plataforma

```python
#@app.get('/max_duration')

def get_max_duration(año,tipo_duracion, plataforma):
    lista = plataformas[(plataformas['release_year'] == año) & (plataformas['duration_type'] == tipo_duracion) & (plataformas['plataforma'] == plataforma)]
    lista = lista.loc[lista['duration_int'] == lista['duration_int'].max()]
    respuesta = lista['title']
    return respuesta
```

```python
get_max_duration(2016,'min','disney')
```

Funcion 2: Cantidad de películas

```python
#@app.get('/cantidad de peliculas')
def get_score_count (plataforma,puntaje,año):
    lista2 = plataformas[(plataformas['plataforma'] == plataforma) & (plataformas['score'] > puntaje) & (plataformas['release_year'] == año)]
    respuesta2 = lista2.shape[0]
    return respuesta2
```

```python
get_score_count('netflix',3.5,2019)
```

Funcion 4: Actor mas recurrente por año y plataforma

```python
#@app.get ('/actor mas recurrente por año y por plataforma ')
#funcion 4: devuelve en nombre del actor por plataforma y por año, más recurrente
def get_actor(plataforma, año):
    result = plataformas[(plataformas['plataforma']==plataforma) & (plataformas['release_year']==año)]
    for i in result['cast']:
        if i != 'Sin dato ':
            i=i.replace(', ' , ',')
        else:
            pass  
    lista4=[]
    for i in result['cast']:
        if i != 'Sin dato':
            s=i.split(',')
            for j in range(len(s)):             
                if s[j] not in lista4:
                    lista4.append(s[j])
                else:
                    pass
        else:
            pass    
    lista4=list(set(lista4))
    contador = 0
    dict={}
    for i in lista4:
        contador = 0
        for j in result['cast']:
            if i in j.split(','):
                contador+=1
        dict[i]=contador
    if len(dict)==0:
        return 'la plataforma no brinda esta informacion'
    else:
        return max(dict,key=dict.get)
```
